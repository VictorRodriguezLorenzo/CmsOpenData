# How to do an Analysis in Particle Physics?

## Introduction

The aim of this **4-hour practical lab** is to investigate muons reconstructed by the CMS detector using a Z boson decay into a muon pair (Z → μμ). Students will explore a real dataset, define muon selection criteria, measure selection efficiencies and extract the mass and width of the Z boson.

The main goals are:

* Learn the workflow of a particle physics data analysis with ROOT and RDataFrame.
* Explore muon kinematics, quality identification and isolation.
* Calculate conditional/cumulative event efficiencies, and per-muon ID and isolation efficiencies versus $p_T$ and $\eta$, including statistical uncertainties.
* Select Z → μμ candidates from backgrounds and measure the mass and width of the Z boson.
* Discuss the efficiency–purity trade-off and limitations of using a preselected dataset.

'RDFAnalyzer' is a set of codes keeping all necessary tools needed to carry out a basic analysis in Particle Physics. It's been developed to be used by junior and senior students in the Physics Bachelor. The input is a **small, preselected subset of CMS 2016 Run G data** at $\sqrt{s}=13$ TeV; it is **not** the full 2024–2025 Run 3 dataset.

## Work tools

The course environment uses **ROOT (version 6.24/06) / PyROOT (python3 extension) and Python (Python 3.9)**, and the RDataFrame interface to CMS NanoAOD ntuples.
The framework is running over docker containers running over the opendata platform at IFCA (https://opendata.ifca.es/). It expects the dataset at the path configured in `RDFAnalyzer/utils.py` and the course workspace at `~/work/CmsOpenData`.

The code exploits the capabilities of ROOT's RDataFrames. RDataFrame offers a modern, high-level interface for analysis of data stored in TTree , CSV and other data formats, in C++ or Python.
Useful references:

* [TEfficiency documentation](https://root.cern.ch/doc/master/classTEfficiency.html)

## Task Description 

Starting with all physical events in our dataset, you will be asked to classify the events following a physical selection criteria, deciding which muons could have been produced in a Z->mumu decay.

You already have some basics about how the detector works, e.g. what kind of 'traces' the different particles leave in the various detectors (tracker, calorimeters, muon chambers) and which are the main kinematics variables in a modern experimental particle physics standard analysis. 

The practical exercise will follow these steps: 

1. Load the data and understand ROOT's ntuple structure.

2. Muon identification and isolation: display different histograms for the main kinematics variables used in the muon identification and isolation. Study the different variables and establish the selection criteria.

3. Candidates selection: Apply the selection criteria from point 2 and study the kinematic variables of the prompt muons you have selected, in particular, the invariant mass. 
 
4. Analysis: Repeat steps 2 and 3 until an optimal selection for Z → µµ events is achieved. 

5. Z boson measurement: find out the invariant mass of opposite changed mu+ mu- pairs and identify the Z → μμ candidates. Do a statistical analysis of the results to measure the Z boson properties (Z mass and decay width).

## Lab exercise

### [Part 0](./notebooks/Part0_Load_dataset.ipynb): Inspect the dataset
* Load ROOT data and explore the `Events` tree and muon NanoAOD branches.

### [Part 1](./notebooks/Part1_Analysis_workflow.ipynb): Analysis workflow
* Plot baseline variables and compare different muon selections.
* Use identification and isolation criteria to select dimuon candidates.
* Save a selected dimuon invariant-mass histogram.

### [Part 2](./notebooks/Part2_Muon_efficiencies.ipynb): Muon selection efficiencies
* Derive step and cumulative cut efficiencies for events with at least two selected muons.
* Compute Loose, Medium and Tight ID efficiencies relative to reconstructed baseline muons.
* Measure Tight ID plus isolation efficiency, including isolation conditional on Tight ID.
* Plot efficiency versus $p_T$ and $\eta$ with `TEfficiency` binomial uncertainties.
* Reflect on biased denominators and the distinction between selection and reconstruction/trigger efficiency.

### [Part 3](./notebooks/Part3_Z_boson_fit.ipynb): Fit the Z peak
* Fit the selected invariant-mass distribution with Gaussian, relativistic Breit–Wigner and convolution functions.
* Compare goodness of fit and discuss the extracted Z mass and width.
